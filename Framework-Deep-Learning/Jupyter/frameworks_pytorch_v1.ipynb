```python
import torch

def print_tensor_info(name, tensor):
    bits = tensor.element_size() * 8
    print(f"{name:10} | Dtype: {str(tensor.dtype):15} | Bits por elemento: {bits}")

t1 = torch.tensor([1, 2, 3], dtype=torch.float32)
t2 = t1.half()     # float16
t3 = t1.bfloat16() # bfloat16
t4 = t1.to(torch.int8)

print_tensor_info("Tensor 1", t1)
print_tensor_info("Tensor 2", t2)
print_tensor_info("Tensor 3", t3)
print_tensor_info("Tensor 4", t4)
```

```python
import torch
from torchviz import make_dot
from IPython.display import display
```

```python
x = torch.randn(2, 3, requires_grad=True)
y = torch.randn(3, 4, requires_grad=True)
z = x @ y
loss = z.sum()

dot = make_dot(loss, params={'x': x, 'y': y})
display(dot)
```

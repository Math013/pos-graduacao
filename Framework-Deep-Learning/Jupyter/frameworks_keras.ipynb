```python
import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import tensorflow as tf
from sklearn import preprocessing
from sklearn.model_selection import train_test_split


os.environ['KERAS_BACKEND'] = 'tensorflow'

import keras
from keras import layers, models, optimizers, losses, metrics
from keras.callbacks import History

print(f"Backend em uso: {keras.backend.backend()}")
print(f"Versão do Keras: {keras.__version__}")
```

## Definindo funções adicionais

```python
def plot_convergence(hist: History, metric: str) -> None:
    """
    Plot training and validation metrics over epochs.

    Parameters
    ----------
    hist : keras.callbacks.History
        History object returned by model.fit().
    metric : str
        Name of the metric to plot (e.g., 'loss', 'accuracy').
    """
    df: pd.DataFrame = pd.DataFrame(hist.history)
    epochs: np.ndarray = np.arange(df[metric].shape[0])

    fig = px.line(
        df,
        x=epochs,
        y=[metric, f"val_{metric}"]
    )
    fig.show()


def reset_seeds(seed: int = 42) -> None:
    """
    Reset random seeds for reproducibility.

    Parameters
    ----------
    seed : int, default=42
        Seed value used across libraries.
    """
    os.environ["PYTHONHASHSEED"] = str(seed)
    tf.random.set_seed(seed)
    np.random.seed(seed)
    random.seed(seed)
    keras.utils.set_random_seed(seed)


print("Funções definidas com sucesso!")
```

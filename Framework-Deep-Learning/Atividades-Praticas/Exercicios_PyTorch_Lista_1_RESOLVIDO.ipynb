# NÍVEL 2: Tipos de Tensores (dtype)

## 📝 EXERCÍCIO 2.1: Diferentes dtypes

**Tarefa:**
Crie um tensor com valor 42 em cada tipo:
- torch.int32
- torch.int64
- torch.float32
- torch.float64
- torch.bool

Para cada um, imprima:
- O tensor
- Seu dtype
- Seu tamanho em bytes

```python
tipos = [torch.int32, torch.int64, torch.float32, torch.float64, torch.bool]

for dt in tipos:
    t = torch.tensor(42, dtype=dt)
    nbytes = t.element_size() * t.numel()
    print(f"{str(dt):15} -> valor={t.item()!s:6} | bytes={nbytes}")
```

## 📝 EXERCÍCIO 2.2: Conversão entre tipos

**Tarefa:**
1. Crie: t1 = torch.tensor([1.5, 2.5, 3.5])
2. Converta para int32: t2 = t1.to(torch.int32)
3. Imprima ambos
4. O que aconteceu?
5. Converta de volta para float64

```python
t1 = torch.tensor([1.5, 2.5, 3.5])
t2 = t1.to(torch.int32)

print("t1:", t1)
print("t2:", t2)

t3 = t2.to(torch.float64)
print("t3:", t3, "| dtype:", t3.dtype)  # informação já foi perdida no truncamento
```

## 📝 EXERCÍCIO 2.3: Verificar tipo

**Tarefa:**
1. Crie 3 tensores com tipos diferentes
2. Para cada um, imprima:
   - t.dtype
   - t.dtype == torch.float32
3. Crie função que retorna "int" ou "float" ou "outro"

```python
ta = torch.tensor([1, 2, 3])
tb = torch.tensor([1.0, 2.0])
tc = torch.tensor([True, False])

for t in (ta, tb, tc):
    print(t.dtype, "| é float32?", t.dtype == torch.float32)

def categoria(t: torch.Tensor) -> str:
    if t.dtype.is_floating_point:
        return "float"
    if t.dtype in (torch.int8, torch.int16, torch.int32, torch.int64, torch.uint8):
        return "int"
    return "outro"

print(categoria(ta), categoria(tb), categoria(tc))  # int float outro
```

# NÍVEL 3: Manipulação de Shape

## 📝 EXERCÍCIO 3.1: Shape e dimensões

**Tarefa:**
1. Crie tensor com shape (2, 3, 4)
2. Imprima:
   - tensor.shape
   - tensor.size()
   - len(tensor.shape)
   - tensor.numel() (total elementos)
   - tensor.dim() (dimensões)

```python
t = torch.zeros(2, 3, 4)

print("shape: ", t.shape)
print("size(): ", t.size())
print("len(shape): ", len(t.shape))
print("numel(): ", t.numel())
print("dim(): ", t.dim())
```

## 📝 EXERCÍCIO 3.2: Reshape

**Tarefa:**
1. Crie: torch.arange(12)
2. Reshape para (3, 4)
3. Reshape para (2, 2, 3)
4. Reshape para (1, 12)
5. Imprima o shape em cada etapa

```python
t = torch.arange(12)
print("original:", t.shape)

a = t.reshape(3, 4)
print("(3,4):   ", a.shape)

b = t.reshape(2, 2, 3)
print("(2,2,3): ", b.shape)

c = t.reshape(1, 12)
print("(1,12):  ", c.shape)
```

## 📝 EXERCÍCIO 3.3: Flatten e Unflatten

**Tarefa:**
1. Crie tensor com shape (3, 4, 2)
2. Use flatten() e view(-1)
3. Qual é a diferença?
4. Converta de volta para (3, 4, 2)

```python
t = torch.arange(24).reshape(3, 4, 2)

f1 = t.flatten()
f2 = t.view(-1)

print("flatten:", f1.shape, "| view(-1):", f2.shape)

back = f1.reshape(3, 4, 2)
print("de volta:", back.shape)
```

## 📝 EXERCÍCIO 3.4: Squeeze e Unsqueeze

**Tarefa:**
1. Crie tensor com shape (3, 1, 4)
2. Use squeeze()
3. Use unsqueeze(0) e unsqueeze(-1)
4. Imprima shape em cada etapa

```python
t = torch.zeros(3, 1, 4)
print("original: ", t.shape)

s = t.squeeze()
print("squeeze(): ", s.shape)

u0 = s.unsqueeze(0)
print("unsqueeze(0):  ", u0.shape)

um1 = s.unsqueeze(-1)
print("unsqueeze(-1): ", um1.shape)
```

# NÍVEL 4: Operações Algébricas

## 📝 EXERCÍCIO 4.1: Operações elemento a elemento

## 📝 EXERCÍCIO 4.5: Produto de Matrizes

**Tarefa:**
1. Crie:
   - A = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
   - B = torch.tensor([[5, 6], [7, 8]], dtype=torch.float32)
2. Calcule: A @ B
3. Qual é o shape?
4. Verifique manualmente

```python
A = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
B = torch.tensor([[5, 6], [7, 8]], dtype=torch.float32)

C = A @ B
print(C)
print("shape:", C.shape)

# Verificação manual:
# C[0,0] = 1*5 + 2*7 = 19 ; C[0,1] = 1*6 + 2*8 = 22
# C[1,0] = 3*5 + 4*7 = 43 ; C[1,1] = 3*6 + 4*8 = 50
assert torch.equal(C, torch.tensor([[19., 22.], [43., 50.]]))
print("✓ conferido manualmente")
```
